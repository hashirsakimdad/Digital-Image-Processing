## 0. Imports & Image Path

Import the required libraries and set the path to the image file in this folder.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import matplotlib.image as mpimg
import os

# Path to the uploaded image in this folder
img_path = "nic ront.jpg"

if not os.path.exists(img_path):
    raise FileNotFoundError(
        f"Image not found: {img_path}. Place your image in this folder and update img_path."
    )

print(f"Using image: {img_path}")
print(f"File size  : {os.path.getsize(img_path)} bytes")
```

## 1–5. Read with OpenCV, Display with Matplotlib, Print Properties

OpenCV reads images in **BGR** order. We convert to **RGB** before showing with Matplotlib so colors look correct.

From `img_cv.shape` we get **height**, **width**, and **channels**. We also print **dtype**, **min**, and **max** pixel values.

```python
# 1) Read image using OpenCV
img_cv = cv2.imread(img_path)

if img_cv is None:
    raise FileNotFoundError(f"Could not read image: {img_path}")

# OpenCV loads as BGR — convert to RGB for correct Matplotlib display
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)

# 2) Display using Matplotlib
plt.figure(figsize=(8, 5))
plt.imshow(img_rgb)
plt.title("Image read with OpenCV, shown with Matplotlib")
plt.axis("off")
plt.show()

# 3) Height, width, number of channels
height, width, channels = img_cv.shape
print(f"Height   : {height}")
print(f"Width    : {width}")
print(f"Channels : {channels}")

# 4) Data type
print(f"Data type: {img_cv.dtype}")

# 5) Minimum (and maximum) pixel value
print(f"Minimum  : {img_cv.min()}")
print(f"Maximum  : {img_cv.max()}")
```

## 6. Access One Pixel

In OpenCV/NumPy arrays, pixels are indexed as `image[row, column]`. OpenCV stores each pixel as **BGR**; the RGB version stores **RGB**.

```python
# 6) Access one pixel (row=100, col=150)
row, col = 100, 150
pixel_bgr = img_cv[row, col]
print(f"Pixel at ({row}, {col}) BGR : {pixel_bgr}")
print(f"  B={pixel_bgr[0]}, G={pixel_bgr[1]}, R={pixel_bgr[2]}")

pixel_rgb = img_rgb[row, col]
print(f"Pixel at ({row}, {col}) RGB : {pixel_rgb}")
```

## 7. Read the Same Image with Pillow

Pillow (`PIL`) returns an `Image` object. Size is `(width, height)` — note this is the opposite order of OpenCV’s `(height, width, channels)`.

```python
# 7) Read the same image using Pillow
img_pil = Image.open(img_path)
print("Pillow image:")
print(f"  Size (W x H): {img_pil.size}")
print(f"  Mode        : {img_pil.mode}")
print(f"  Format      : {img_pil.format}")

plt.figure(figsize=(8, 5))
plt.imshow(img_pil)
plt.title("Image read with Pillow")
plt.axis("off")
plt.show()
```

## 8. Read the Image with Matplotlib

`matplotlib.image.imread` loads the image directly as an RGB NumPy array (values often as float in `[0, 1]` for some formats, or `uint8` for JPEG).

```python
# 8) Read the image using Matplotlib
img_mpl = mpimg.imread(img_path)

print("Matplotlib image:")
print(f"  Shape    : {img_mpl.shape}")
print(f"  Data type: {img_mpl.dtype}")
print(f"  Min      : {img_mpl.min()}")
print(f"  Max      : {img_mpl.max()}")

plt.figure(figsize=(8, 5))
plt.imshow(img_mpl)
plt.title("Image read with Matplotlib")
plt.axis("off")
plt.show()
```

## 9. Save the Image with OpenCV

`cv2.imwrite` expects a **BGR** array. We save a copy as `opencv_saved.jpg`.

```python
# 9) Save the image using OpenCV
opencv_out = "opencv_saved.jpg"
ok = cv2.imwrite(opencv_out, img_cv)

if ok:
    print(f"Saved with OpenCV -> {opencv_out}")
else:
    print("OpenCV failed to save the image.")
```

## 10. Save the Image with Pillow

Use `Image.save()` to write a Pillow image. We save a copy as `pillow_saved.jpg`.

```python
# 10) Save the image using Pillow
pillow_out = "pillow_saved.jpg"
img_pil.save(pillow_out)
print(f"Saved with Pillow -> {pillow_out}")

print("\nDone. Check this folder for:")
print(f"  - {opencv_out}")
print(f"  - {pillow_out}")
```
